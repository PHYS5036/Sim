# Geant4 geometry and tracks in Jupyter

First run `bin/visualize.py` as described in the README. This notebook needs only Plotly, not Geant4 or ROOT. Rotate by dragging, zoom by scrolling, and click legend entries to hide volumes or particle groups. Mesh coordinates and track points are in mm.

The default GammaLabMC sample is a forward-directed monoenergetic illustration, **not** an absolute-efficiency simulation. VRML retains track colours and paths, but not event IDs, track IDs, energies or PDG codes. Particle labels here follow the colour mapping in the supplied visualization macro.

```python
import os
import sys
from pathlib import Path

repo = Path.cwd() if (Path.cwd() / 'bin/geant4_scene.py').exists() else Path.cwd().parent
sys.path.insert(0, str(repo / 'bin'))
from geant4_scene import load_scene, figure

scene_path = Path(os.environ.get('GEANT4_SCENE', str(repo / 'visualization/scene.json')))
scene = load_scene(scene_path)
print(f"{len(scene['geometry'])} volumes, {len(scene['tracks'])} track polylines; units: {scene['units']}")
fig = figure(scene, geometry_opacity=0.25)
fig.show(renderer='notebook')
```

The `notebook` renderer embeds Plotly JavaScript; no CDN is needed. Trust the notebook to allow JavaScript output. For JupyterLab or VS Code, use `fig.show(renderer='plotly_mimetype')` instead. The generated `scene.html` is also an offline, rotatable viewer.

To inspect a volume numerically: `scene['geometry'][0]` contains its name, vertices, triangle indices and colour. `scene['tracks'][0]` contains a polyline's points and colour.
